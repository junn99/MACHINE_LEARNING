## 문자형 - 날짜형 데이터 변환

### 날짜 -> 문자

```python
s = datetime(2011,1,3)
print(s,type(s))

s1=str(s)
print(s1,type(s1))

s2 = s.strftime("%Y-%m-%d")
print(s2,type(s2))
```

```python
datetime.strftime?
    # datetime을 str로 변환
```

### 문자 -> 날짜

```python
s = "2011-01-03"

s = datetime.strptime(s,"%Y-%m-%d")
s
```

```python
import pandas as pd

datestrs = ["2011-07-06 12:00:00", "2011-08-06 00:00:00"]
pd.to_datetime(datestrs)
```

```python
idx = pd.to_datetime(datestrs + [None])
idx
print(idx[2])
    # NaT(Not a time) : 판다스에서 누락된 타임스탬프 데이터를 나타냄
pd.isna(idx)
```
