## 전처리

### 결측치

```python
# 결측치 비율 확인
data.isna().mean()
```

결측치 채우기
- 데이터에 등장하지 않을 법한 임의의 수를 넣음
- 선형모델에선 아웃라이어로 취급되지만
- 트리모델에선 결측치라는 사실 자체로서 유의미한 노드가 생길 것

```python
data = data.dropna(subset=['pref_o_attractive',
'pref_o_sincere',                  
'pref_o_intelligence',            
'pref_o_funny',                     
'pref_o_ambitious',             
'pref_o_shared_interests',          
'attractive_o',                     
'sincere_o',                        
'intelligence_o',                   
'funny_o',                          
'ambitous_o',                
'shared_interests_o'])
    # 중요도와 관련된 부분은 결측치 제거함

data = data.fillna(-99)
data.info() # 결측치 다 사라짐
data.head(10)
```

### 피쳐 엔지니어링

#### 나이

```python
def age_gap(x):
    if x['age'] == -99:
        return -99
    elif x['age_o'] == -99:
        return -99
    elif x['gender'] == 'female':
        return x['age_o'] - x['age']
    else:
        return x['age'] - x['age_o']
```

```python
data['age_gap'] = data.apply(age_gap, axis=1)
```

```python
data['age_gap_abs'] = abs(data['age_gap'])
data['age_gap_abs'].describe()
```

#### 인종

```python
def same_race(x):
    if x['race'] == -99:
        return -99
    elif x['race_o'] == -99:
        return -99
    elif x['race'] == x['race_o']:
        return 1
    else:
        return -1
```

```python
data['same_race'] = data.apply(same_race, axis=1)
data.head()
```

```python
def smae_race_point(x):
    if x['same_race'] == -99:
        return -99
    else:
        return x['same_race'] * x['importance_same_race']
```

```python
data['same_race_point'] = data.apply(smae_race_point, axis=1)
data.head()
```

#### 평가 / 중요도

```python
def rating(data, importance, score):
    if data[importance] == -99:
        return -99
    elif data[score] == -99:
        return -99
    else:
        return data[importance] * data[score]
```

```python
partner_imp = data.columns[8:14] # 상대방의 중요도
partner_rate_me = data.columns[14:20] # 본인에 대한 상대방의 평가
my_imp = data.columns[20:26]
my_rate_partner = data.columns[26:32]
```

```python
#상대방 관련 새 변수 이름을 저장하는 리스트
new_label_partner = ['attractive_p', 'sincere_p', 'intelligence_p',\
                    'funny_p','ambition_p','shared_interests_p']

#본인 관련 새 변수 이름을 저장하는 리스트
new_label_me = ['attractive_m', 'sincere_m', 'intelligence_m',\
                    'funny_m','ambition_m','shared_interests_m']
```

```python
for i,j,k in zip(new_label_partner, partner_imp, partner_rate_me):
    data[i] = data.apply(lambda x: rating(x, j, k), axis = 1)
```

```python
for i,j,k in zip(new_label_me, my_imp, my_rate_partner):
    data[i] = data.apply(lambda x: rating(x, j, k), axis = 1)
#rating 함수 적용(본인이 파트너를 평가했을 때,)
```

```python
data = pd.get_dummies(data, columns=['gender','race','race_o'], drop_first=True)
```

## 모델링 & 평가

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(data.drop('match', axis=1),
                                                    data['match'],test_size=0.2,
                                                    random_state=100)
```

```python
import xgboost as xgb
```

```python
model = xgb.XGBClassifier(n_estimators=500, max_depth=5,random_state=100)
```

```python
model.fit(X_train,y_train)
```

```python
pred = model.predict(X_test)
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
accuracy_score(y_test, pred)
```

```python
print(confusion_matrix(y_test,pred))
```

```python
print(classification_report(y_test, pred))
```
